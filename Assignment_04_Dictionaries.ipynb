## Q1. Basic Dictionary Creation

```python
student = {
    "name": "Muhammad Haris",
    "age": 17,
    "course": "ICS",
    "city": "Faisalabad"
}

print(student["name"])
print(student["age"])
print(student["course"])
print(student["city"])
```

## Q2. Adding and Updating Values

```python
car = {
    "brand": "Toyota",
    "model": "Corolla",
    "year": 2018
}

car["color"] = "White"
car["year"] = 2022

print(car)
```

## Q3. Loop Through Dictionary

```python
person = {
    "name": "Ali",
    "age": 25,
    "city": "Karachi"
}

for key in person:
    print(key, ":", person[key])
```

## Q4. Count Frequency of Numbers

```python
numbers = [1, 2, 2, 3, 3, 3, 4, 4, 4, 4]

frequency = {}

for number in numbers:
    if number in frequency:
        frequency[number] = frequency[number] + 1
    else:
        frequency[number] = 1

print(frequency)
```

## Q5. Word Counter

```python
sentence = input("Enter a sentence: ")

words = sentence.split()
frequency = {}

for word in words:
    if word in frequency:
        frequency[word] = frequency[word] + 1
    else:
        frequency[word] = 1

print(frequency)
```

## Q6. Student Marks and Average

```python
marks = {
    "Math": 80,
    "English": 70,
    "Physics": 90
}

total = 0

for subject in marks:
    total = total + marks[subject]

average = total / len(marks)

print("Total =", total)
print("Average =", average)
```

## Q7. Highest Scoring Student

```python
students = {
    "Ali": 85,
    "Sara": 92,
    "Ahmed": 78,
    "Zara": 95
}

highest_student = ""
highest_marks = 0

for name in students:
    if students[name] > highest_marks:
        highest_marks = students[name]
        highest_student = name

print("Highest scoring student:", highest_student)
print("Marks:", highest_marks)
```

## Q8. Merge Dictionaries

```python
dict1 = {
    "a": 1,
    "b": 2
}

dict2 = {
    "c": 3,
    "d": 4
}

merged = dict1.copy()
merged.update(dict2)

print(merged)
```

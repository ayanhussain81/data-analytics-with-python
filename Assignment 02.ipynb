1) A company decided to give bonus of 5% to employee if his/her year of service is more than 5 years.

12) A school has following rules for grading system:

a. Below 25 - F

b. 25 to 45 - E

c. 45 to 50 - D

d. 50 to 60 - C

e. 60 to 80 - B

f. Above 80 - A

Ask user to enter marks and print the corresponding grade.

```python
#Type your code here
def get_grade(score):
    if score >= 90:
        return "A"
    elif score >= 80:
        return "A"
    elif score >= 70:
        return "B"
    elif score >= 60:
        return "B"
    elif score >= 50:
        return "B"
    elif score >= 45:
        return "D"
    elif score >= 25:
        return "E"
    else:
        return "F"

scor = get_grade(int(input("Enter your score: ")))
print(f"Your grade is {scor}.")
```

14)A student will not be allowed to sit in exam if his/her attendence is less than 75%.

Take following input from user

- Number of classes held

- Number of classes attended.

And print

- percentage of class attended

- Is student is allowed to sit in exam or not.


```python
#Type your code here

classes = int(input("Enter the number of classes held: "))

attended = int(input("Enter the number of classes attended: "))

if attended > classes:
    print("Invalid input! Attended classes cannot be greater than classes held.")
else:
    attendance = attended / classes * 100

    print(f"Total Attendance percentage is {int(attendance)}%")

    if attendance >= 75:
        print(f"You have attended {attended} classes out of {classes}. You are eligible to sit in the exam.")
    else:
        print(f"You have attended {attended} classes out of {classes}. You are not eligible to sit in the exam.")
```

16) Write a program to check if a year is leap year or not.

If a year is divisible by 4 then it is leap year but if the year is century year like 2000, 1900, 2100 then it must be divisible by 400.

```python
#Type your code here
year = int(input("Enter the year: "))

if year % 4 == 0:
    if year % 100 == 0:
        if year % 400 == 0:
            print(f"{year} is a leap year.")
        else:
            print(f"{year} is not a leap year.")
    else:
        print(f"{year} is a leap year.")
else:
    print(f"{year} is not a leap year.")
```

6) Write a program to calculate the electricity bill (accept number of unit from user) according to the following criteria :
     Unit                                                     Price  
uptp 100 units                                             no charge
Next 200 units                                              Rs 5 per unit
After 200 units                                             Rs 10 per unit
(For example if input unit is 350 than total bill amount is Rs.3500
(For example if input unit is 97 than total bill amount is Rs.0
(For example if input unit is 150 than total bill amount is Rs.750

```python
units = int(input("Enter the number of units: "))

if units <= 100:
    bill = 0
elif units <= 300:
    bill = (units - 100) * 5
else:
    bill = 200 * 5 + (units - 300) * 10

print(f"Total bill amount is Rs. {bill}")
```

```python
 # All python operators
a = 10
b = 3

print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a // b)
print(a % b)
print(a ** b)

print(a > b)
print(a < b)
print(a == b)
print(a != b)
print(a >= b)
print(a <= b)

print(a > 5 and b < 5)
print(a > 5 or b > 5)
print(not(a > b))

a += 2
print(a)
a -= 2
print(a)
a *= 2
print(a)
a /= 2
print(a)

x = [1, 2, 3]
print(2 in x)
print(5 not in x)

c = x
print(x is c)
print(x is not c)

print(5 & 3)
print(5 | 3)
print(5 ^ 3)
print(~5)
print(5 << 1)
print(5 >> 1)
```

```python
#Marks and Percentage
english = 85
islamiat = 90
maths = 80

total_marks = 300
obtained_marks = english + islamiat + maths

percentage = (obtained_marks / total_marks) * 100

print("English:", english)
print("Islamiat:", islamiat)
print("Maths:", maths)
print("Total Marks:", total_marks)
print("Obtained Marks:", obtained_marks)
print("Percentage:", percentage)
```

```python
#Swap two variables without a third variable
a = 10
b = 20

print("Before swapping:")
print("a =", a)
print("b =", b)

a, b = b, a

print("After swapping:")
print("a =", a)
print("b =", b)
```

```python
#Discount and Final Price
price = 5000
discount_percentage = 20

discount_amount = (price * discount_percentage) / 100
final_price = price - discount_amount

print("Price:", price)
print("Discount Percentage:", discount_percentage)
print("Discount Amount:", discount_amount)
print("Final Price:", final_price)
```

```python
# Circumference and Area of Circle

pi = 3.14159
radius = float(input("Enter radius: "))

area = pi * radius ** 2
circumference = 2 * pi * radius

print("Area =", area)
print("Circumference =", circumference)
```

## 1. Print alternate elements of a list

```python
numbers = input("Enter numbers separated by spaces: ").split()

for i in range(0, len(numbers), 2):
    print(numbers[i])
```

## 2. Reverse a list without using `reverse()`

```python
numbers = input("Enter numbers separated by spaces: ").split()

reversed_list = []

for i in range(len(numbers) - 1, -1, -1):
    reversed_list.append(numbers[i])

print("Reversed list:", reversed_list)
```

## 3. Find the largest number without using `max()`

```python
numbers = input("Enter numbers separated by spaces: ").split()

numbers = [int(x) for x in numbers]

largest = numbers[0]

for number in numbers:
    if number > largest:
        largest = number

print("Largest number:", largest)
```

## 4. Rotate the elements of a list

```python
numbers = input("Enter numbers separated by spaces: ").split()

last = numbers[-1]

for i in range(len(numbers) - 1, 0, -1):
    numbers[i] = numbers[i - 1]

numbers[0] = last

print("Rotated list:", numbers)
```

## 5. Delete a given word from a string

```python
sentence = input("Enter a string: ")
word = input("Enter the word to delete: ")

words = sentence.split()
new_words = []

for item in words:
    if item != word:
        new_words.append(item)

sentence = " ".join(new_words)

print("New string:", sentence)
```

## 6. Convert a date from mm/dd/yyyy to Month day, year

```python
date = input("Enter date in mm/dd/yyyy format: ")

month, day, year = date.split("/")

months = [
    "January", "February", "March", "April",
    "May", "June", "July", "August",
    "September", "October", "November", "December"
]

month_name = months[int(month) - 1]

print(month_name, int(day), ",", year)
```

## 7. Capitalize the first character of each word

```python
def capitalize_words(sentence):
    words = sentence.split()
    new_words = []

    for word in words:
        new_words.append(word.capitalize())

    return " ".join(new_words)

sentence = input("Enter a sentence: ")

print(capitalize_words(sentence))
```

## 8. Find the sum of each row of a matrix

```python
matrix = [
    [10, 12, 10],
    [10, 11, 10],
    [20, 20, 23]
]

for i in range(len(matrix)):
    total = 0

    for number in matrix[i]:
        total = total + number

    print("Sum of row", i + 1, "=", total)
```

## 9. Add two matrices

```python
matrix1 = [
    [1, 2, 3],
    [4, 5, 6]
]

matrix2 = [
    [7, 8, 9],
    [10, 11, 12]
]

result = []

for i in range(len(matrix1)):
    row = []

    for j in range(len(matrix1[i])):
        row.append(matrix1[i][j] + matrix2[i][j])

    result.append(row)

print("Sum of matrices:")

for row in result:
    print(row)
```

## 10. Multiply two matrices

```python
matrix1 = [
    [1, 2],
    [3, 4]
]

matrix2 = [
    [5, 6],
    [7, 8]
]

result = [
    [0, 0],
    [0, 0]
]

for i in range(len(matrix1)):
    for j in range(len(matrix2[0])):
        for k in range(len(matrix2)):
            result[i][j] = result[i][j] + matrix1[i][k] * matrix2[k][j]

print("Product of matrices:")

for row in result:
    print(row)
```

1. Write a program that accepts a list from user and print the alternate element of list. 

```python
num = int(input("How many numbers do you want to enter: "))

num_list = []

for i in range(num):
    value = int(input("Enter number: "))
    num_list.append(value)
print(num_list)

print("Alternate elements:")

for i in range(0, len(num_list), 2):
    print(num_list[i])

```

3. Find and display the largest number of a list without using built-in function max(). Your program
should ask the user to input values in list from keyboard. 

```python
num = int(input("How many numbers do you want to enter: "))

num_list = []

for i in range(num):
    value = int(input("Enter number: "))
    num_list.append(value)
print("user list:",  num_list)

largest = num_list[0]

for i in range(len(num_list)):
    if num_list[i] > largest:
        largest = num_list[i]
print("Largest number:", largest)
```

4. Write a program that rotates the element of a list so that the element at the first index moves to the
second index, the element in the second index moves to the third index, etc., and the element in the last
index moves to the first index. 


```python
num_list = [2, 4, 6, 8, 10]

new_list = []

new_list.append(num_list[4])

for i in range(0, len(num_list) - 1):
    new_list.append(num_list[i])

print("Original list:", num_list)
print("Rotated list:", new_list)
```

6. Write a program that reads a string from the user containing a date in the form mm/dd/yyyy. It
should print the date in the form March 12, 2021. 

```python
Date = input("Enter a date in mm/dd/yyyy format: ")

Date = Date.split("/")

month = Date[0]
day = Date[1]
year = Date[2]

if month == "01":
    month = "January"
elif month == "02":
    month = "February"
elif month == "03":
    month = "March"
elif month == "04":
    month = "April"
elif month == "05":
    month = "May"
elif month == "06":
    month = "June"
elif month == "07":
    month = "July"
elif month == "08":
    month = "August"
elif month == "09":
    month = "September"
elif month == "10":
    month = "October"
elif month == "11":
    month = "November"
elif month == "12":
    month = "December"

print(month, day + ",", year)
```

8. Find the sum of each row of matrix of size m x n. 

```python
matrix = [
    [4, 1, 6],
    [3, 8, 2],
    [9, 5, 7]
]


m = len(matrix)

for i in range(m):
    row_sum = 0  
    for j in range(len(matrix[i])):
        row_sum += matrix[i][j]
    
    print(f"Sum of Row {i + 1} = {row_sum}")
```

9. Write a program to add two matrices of size n x m. 

```python

A = [
    [1, 3, 5],
    [2, 4, 6],
    [7, 8, 9]
]


B = [
    [9, 7, 5],
    [8, 6, 4],
    [3, 2, 1]
]


n = len(A)
m = len(A[0])


C = [[0 for _ in range(m)] for _ in range(n)]

for i in range(n):
    for j in range(m):
        C[i][j] = A[i][j] + B[i][j]


print("Resultant Matrix (A + B):")
for row in C:
    print(row)
```

10. Write a program to multiply two matrices 

```python

A = [
    [1, 2, 3],
    [4, 5, 6]
]


B = [
    [7, 8],
    [9, 1],
    [2, 3]
]


rows_A = len(A)
cols_A = len(A[0])
cols_B = len(B[0])


C = [[0 for _ in range(cols_B)] for _ in range(rows_A)]


for i in range(rows_A):          
    for j in range(cols_B):      
        for k in range(cols_A):  
            C[i][j] += A[i][k] * B[k][j]


print("Resultant Matrix (A x B):")
for row in C:
    print(row)
```

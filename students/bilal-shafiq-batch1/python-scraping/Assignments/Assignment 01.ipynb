3. Write a program in which use all the operators we can use in Python

```python
a = 10
b = 3

# Arithmetic operators
print("Addition (a + b):", a + b)
print("Subtraction (a - b):", a - b)
print("Multiplication (a * b):", a * b)
print("Division (a / b):", a / b)
print("Floor Division (a // b):", a // b)
print("Modulus (a % b):", a % b)
print("Exponent (a ** b):", a ** b)

# Comparison operators
print("\nEqual (a == b):", a == b)
print("Not Equal (a != b):", a != b)
print("Greater Than (a > b):", a > b)
print("Less Than (a < b):", a < b)
print("Greater or Equal (a >= b):", a >= b)
print("Less or Equal (a <= b):", a <= b)

# Logical operators
x = True
y = False
print("\nAND (x and y):", x and y)
print("OR (x or y):", x or y)
print("NOT (not x):", not x)

# Assignment operators
c = 5
c += 2; print("\nc += 2 ->", c)
c -= 2; print("c -= 2 ->", c)
c *= 3; print("c *= 3 ->", c)
c /= 2; print("c /= 2 ->", c)
c //= 2; print("c //= 2 ->", c)
c **= 2; print("c **= 2 ->", c)
c %= 3; print("c %= 3 ->", c)

# Bitwise operators
print("\nAND (a & b):", a & b)
print("OR (a | b):", a | b)
print("XOR (a ^ b):", a ^ b)
print("NOT (~a):", ~a)
print("Left Shift (a << 1):", a << 1)
print("Right Shift (a >> 1):", a >> 1)

# Identity operators
p = [1, 2, 3]
q = [1, 2, 3]
r = p
print("\nis (p is r):", p is r)
print("is not (p is not q):", p is not q)

# Membership operators
lst = [1, 2, 3, 4, 5]
print("\nin (3 in lst):", 3 in lst)
print("not in (10 not in lst):", 10 not in lst)
```

4. Completes the following steps of small task:
    - Mention Marks of English , Islamiat and Maths out of 100 in 3 different variables
    - Mention Variable of Total Marks and assign 300 to it
    - Calculate Percentage

```python
english_marks = 85
islamiat_marks = 78
maths_marks = 92

total_marks = 300
obtained_marks = english_marks + islamiat_marks + maths_marks

percentage = (obtained_marks / total_marks) * 100

print("English Marks:", english_marks)
print("Islamiat Marks:", islamiat_marks)
print("Maths Marks:", maths_marks)
print("Total Marks:", total_marks)
print("Obtained Marks:", obtained_marks)
print("Percentage:", percentage, "%")
```

6. Take the radius of a circle from a variable and calculate its Area and Circumference (use `pi = 3.14159`)

```python
pi = 3.14159
radius = 7

area = pi * (radius ** 2)
circumference = 2 * pi * radius

print("Radius:", radius)
print("Area:", area)
print("Circumference:", circumference)
```

7. Take the price of an item and a discount percentage in variables, then calculate and print the discount amount and the final price after discount

```python
price = 1500
discount_percent = 20

discount_amount = (price * discount_percent) / 100
final_price = price - discount_amount

print("Original Price:", price)
print("Discount Percentage:", discount_percent, "%")
print("Discount Amount:", discount_amount)
print("Final Price:", final_price)
```

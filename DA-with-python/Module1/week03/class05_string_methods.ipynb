# list Slicing

```python
cities = ["Karachi", "Lahore", "Islamabad", "Peshawar", "Quetta"]

print(cities[0:3])  #   [start:stop] [included : excluded]
print(cities[0:4])
print(cities[1:4])
print(cities[1:5])
print(cities[3:5])
```

```python
print(cities[:3]) 
print(cities[3:])
print(cities[:])
```

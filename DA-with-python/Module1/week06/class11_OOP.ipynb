A class is a blueprint. It describes what a thing holds (its attributes) and what it can do (its methods). An instance is one specific thing built from that blueprint.



```python
# data --> attributes 
# functionality --> methods
```

```python
class Student:

    def __init__(self, name, age,score):
        self.name = name
        self.age = age
        self.score = score


# __init__ --> Constructor  no need to call -- will run automatically
# def __init__ — "when a new Student is created, do this:"
# self — "the new object itself"
# self.name = name — "store name on this object"

# object creation 
s3 = Student("ryan",50,836)
print(s3.name)
print(s3.age)
print(s3.score)

# Student.__init__(s3, "Ayan",50,836 )
#   s3.name = name
#   s3.age = age
#   s3.score =score
```

```python
class Student:

    def __init__(self, name, age,score): # data - attibutes
        self.name = name
        self.age = age
        self.score = score

    # methods 

    def get_grade(self):
        if self.score >= 90:
            return "A Grade"
        elif self.score >= 80:
            return "B Grade"
        else:
            return "C Grade"

Student_4 = Student("Mahad", 10, 95)
print(Student_4.name)

print(Student_4.get_grade())

```

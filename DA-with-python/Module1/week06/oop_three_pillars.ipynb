## PILLAR 2 — Encapsulation 🔒

**Encapsulation** means bundling data and methods together — and **controlling who can access the data.**

### Access Levels in Python

| Style | Level | Meaning |
|---|---|---|
| `name` | Public | Anyone can read or change |
| `_name` | Protected | Internal use — convention only |
| `__name` | Private | Hidden — cannot be accessed directly from outside |

### 2.1 — Public, Protected & Private

````python
class Person:
    def __init__(self, name, age, salary):
        self.name     = name        # public
        self._age     = age         # protected
        self.__salary = salary      # private

    def show_info(self):
        print(f"Name   : {self.name}")
        print(f"Age    : {self._age}")
        print(f"Salary : Rs {self.__salary:,}")

p = Person("Ali", 22, 80000)
p.show_info()
````

````python
print(p.name)   # ✅ public works
print(p._age)   # ⚠️ protected works but bad practice
print(p.__salary) # private object not accessible out of the class
````

````python
# Try accessing private — this will raise AttributeError
try:
    print(p.__salary)
except AttributeError as e:
    print(f"Error: {e}")
````

### 2.2 — Getters & Setters

````python
class Student:
    def __init__(self, name, score):
        self.name    = name
        self.__score = score        # private

    def get_score(self):            # getter — read
        return self.__score

    def set_score(self, score):     # setter — write with validation
        if 0 <= score <= 100:
            self.__score = score
            print(f"Score updated to {score}")
        else:
            print("Score must be between 0 and 100.")

    def __str__(self):
        return f"{self.name} | Score: {self.__score}"

s = Student("Sara", 85)
print(s)
print(s.__score)
print(s.get_score())
s.set_score(95)
s.set_score(150)
print(s)
````

### 2.3 — @property — The Pythonic Way

@ - Decorator

`@property` makes getters and setters look like normal attributes from outside, but runs validation code behind the scenes.

````python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner     = owner
        self.__balance = balance

    @property
    def balance(self):              # getter — read like attribute
        return self.__balance

    @balance.setter
    def balance(self, amount):      # setter — validated on assignment
        if amount >= 0:
            self.__balance = amount
        else:
            print("Balance cannot be negative.")

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount # self.__balance =  self.__balance + amount
            print(f"Deposited Rs {amount:,} → Balance: Rs {self.__balance:,}")

    def withdraw(self, amount):
        if amount > self.__balance:
            print("Insufficient funds.")
        else:
            self.__balance -= amount
            print(f"Withdrew Rs {amount:,} → Balance: Rs {self.__balance:,}")   

    def __str__(self):
        return f"Account({self.owner}) | Balance: Rs {self.__balance:,}"

acc = BankAccount("Ali", 10000)
print(acc)
acc.deposit(5000)
acc.withdraw(3000)
print(f"Balance: {acc.balance}")   # getter
acc.balance = 20000                 # setter
acc.balance = -100                  # rejected
print(acc)
````

## PILLAR 3 — Polymorphism 🔄

**Polymorphism** = same method name, different behaviour depending on the object.

`poly` = many, `morph` = form → **many forms**

> Same word `speak()` — Dog says Woof, Cat says Meow, Cow says Moo.

### 3.1 — Method Overriding

````python
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        print(f"{self.name} makes a sound.")


class Dog(Animal):
    def speak(self):
        print(f"{self.name}: Woof! 🐶")


class Cat(Animal):
    def speak(self):
        print(f"{self.name}: Meow! 🐱")


class Cow(Animal):
    def speak(self):
        print(f"{self.name}: Moo! 🐄")


animals = [Dog("Bruno"), Cat("Luna"), Cow("Bessie")]

for animal in animals:
    animal.speak()   # same call — different output
````

### 3.2 — Duck Typing

> *"If it walks like a duck and quacks like a duck — it's a duck."*

Python doesn't care about the **type** of an object — only that it has the **method** you need.

````python
class Robot:
    def speak(self):
        print("BEEP BOOP. I AM ROBOT. 🤖")

class Human:
    def speak(self):
        print("Hello, how are you? 👋")

# No shared parent — just needs speak()
speakers = [Dog("Rex"), Cat("Kitty"), Robot(), Human()]

for s in speakers:
    s.speak()   # Python doesn't check type — just calls the method
````

### 3.3 — Polymorphism with Shapes

````python
class Shape:
    def __init__(self, color):
        self.color = color

    def area(self):
        return 0

    def describe(self):
        print(f"{self.__class__.__name__:<12} | Color: {self.color:<6} | Area: {self.area():.2f}")


class Circle(Shape):
    def __init__(self, color, radius):
        super().__init__(color)
        self.radius = radius

    def area(self):
        return 3.14159 * self.radius ** 2 # pie * r**2


class Rectangle(Shape):
    def __init__(self, color, width, height):
        super().__init__(color)
        self.width  = width
        self.height = height

    def area(self):
        return self.width * self.height


class Triangle(Shape):
    def __init__(self, color, base, height):
        super().__init__(color)
        self.base   = base
        self.height = height

    def area(self):
        return 0.5 * self.base * self.height


shapes = [
    Circle("red",    5),
    Rectangle("blue", 4, 6),
    Triangle("green", 3, 8),
]

for shape in shapes:
    shape.describe()

total = sum(shape.area() for shape in shapes)
print(f"\nTotal area of all shapes: {total:.2f}")
````

## PILLAR 4 — Abstraction 🧩

**Abstraction** = hide complexity, show only what's necessary.

In OOP: define a **blueprint** that forces every child class to implement certain methods.

**Real-world analogy:**
You press the accelerator → car moves.
You don't know about pistons, fuel injection, combustion.
The complexity is hidden — you only see the simple interface.

### 4.1 — Abstract Class Basics

````python
from abc import ABC, abstractmethod

class Vehicle(ABC):            # abstract class — cannot be instantiated

    def __init__(self, brand):
        self.brand = brand

    @abstractmethod
    def start(self):           # MUST be implemented in every child
        pass

    @abstractmethod
    def stop(self):            # MUST be implemented in every child
        pass

    def honk(self):            # regular method — inherited as-is
        print(f"{self.brand}: Beep beep! 📯")




# Try to instantiate abstract class — will fail
try:
    v = Vehicle("Test")
except TypeError as e:
    print(f"Error: {e}")
````

````python
class Car(Vehicle):
    def start(self):
        print(f"{self.brand} car engine started 🚗")

    def stop(self):
        print(f"{self.brand} car engine stopped.")


class Bike(Vehicle):
    def start(self):
        print(f"{self.brand} bike kick-started 🏍️")

    def stop(self):
        print(f"{self.brand} bike brakes applied.")


class Truck(Vehicle):
    def start(self):
        print(f"{self.brand} truck diesel engine roaring 🚛")

    def stop(self):
        print(f"{self.brand} truck air brakes engaged.")


vehicles = [Car("Toyota"), Bike("Honda"), Truck("Volvo")]

for v in vehicles:
    v.start()
    v.honk()
    v.stop()
    print()
````

### 4.2 — What Happens If a Child Forgets a Method?

````python
class IncompleteVehicle(Vehicle):
    def start(self):
        print("Started!")
    # forgot stop()!

try:
    iv = IncompleteVehicle("Broken")
except TypeError as e:
    print(f"Error: {e}")
````

### 4.3 — Real Example: Payment System

````python
class Payment(ABC):

    def __init__(self, sender):
        self.__sender = sender      # Encapsulation inside Abstraction

    @property # getter
    def sender(self):
        return self.__sender

    @abstractmethod
    def pay(self, amount):
        pass

    @abstractmethod
    def refund(self, amount):
        pass

    def receipt(self, amount):      # shared — not abstract
        print(f"  Receipt: Rs {amount:,} via {self.__class__.__name__}")


class CashPayment(Payment):
    def pay(self, amount):
        print(f"💵 Paying Rs {amount:,} in cash.")

    def refund(self, amount):
        print(f"💵 Refunding Rs {amount:,} in cash.")


class CardPayment(Payment):
    def __init__(self, sender, last4):
        super().__init__(sender)
        self.last4 = last4

    def pay(self, amount):
        print(f"💳 Charging Rs {amount:,} to card ending {self.last4}.")

    def refund(self, amount):
        print(f"💳 Refunding Rs {amount:,} to card ending {self.last4}.")


class MobilePayment(Payment):
    def __init__(self, sender, phone):
        super().__init__(sender)
        self.phone = phone

    def pay(self, amount):
        print(f"📱 Paying Rs {amount:,} via {self.phone}.")

    def refund(self, amount):
        print(f"📱 Refunding Rs {amount:,} to {self.phone}.")


payments = [
    CashPayment("Ali"),
    CardPayment("Sara",  "3456"),
    MobilePayment("Ahmed","0300-1234567"),
]

for p in payments:
    p.pay(5000)
    p.receipt(5000)
    p.refund(1000)
    print()
````

## All 3 Pillars Together 🏆

**Notification System** — uses Encapsulation + Abstraction + Polymorphism in one program.

````python
class Notification(ABC):

    def __init__(self, sender):
        self.__sender = sender           # ENCAPSULATION — private

    @property
    def sender(self):
        return self.__sender

    @abstractmethod
    def send(self, message):             # ABSTRACTION — must implement
        pass

    def log(self, message):
        print(f"  [LOG] {self.__class__.__name__} | {message[:35]}...")


class EmailNotification(Notification):
    def __init__(self, sender, email):
        super().__init__(sender)
        self.email = email

    def send(self, message):             # POLYMORPHISM — own version
        print(f"📧 Email → {self.email}: {message}")


class SMSNotification(Notification):
    def __init__(self, sender, phone):
        super().__init__(sender)
        self.phone = phone

    def send(self, message):             # POLYMORPHISM — own version
        print(f"📱 SMS → {self.phone}: {message}")


class PushNotification(Notification):
    def __init__(self, sender, device):
        super().__init__(sender)
        self.device = device

    def send(self, message):             # POLYMORPHISM — own version
        print(f"🔔 Push → {self.device}: {message}")


msg = "Your order has been shipped and is on its way!"

notifications = [
    EmailNotification("System", "ali@mail.com"),
    SMSNotification("System",   "0300-1234567"),
    PushNotification("System",  "DEVICE-001"),
]

for n in notifications:
    n.send(msg)
    n.log(msg)
    print()
````

## 🔑 Key Takeaways

| Pillar | Rule |
|---|---|
| **Encapsulation** | `__private` → hide data, expose through methods/`@property` |
| **Polymorphism** | Same method name → different body in each class |
| **Abstraction** | `ABC` + `@abstractmethod` → force children to implement |

> The goal of OOP: write code that is **safe**, **flexible**, and **easy to extend**.
